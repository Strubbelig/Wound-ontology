# BFO Reference Generation

One-time notebook to fetch the Basic Formal Ontology (BFO) class hierarchy
and store it as a stable reference CSV at `data/reference/bfo_hierarchy.csv`.

This file is unversioned (no timestamp) and used as a predefined concept seed
for the `ClassHierarchy` in the main pipeline.

```python
from wound_ontology.data_manager.data_manager import DataManager
from wound_ontology.connectors.bfo_connector import BFOConnector
```

```python
dm = DataManager()
bfo = BFOConnector(dm)
```

```python
# Fetch labels (loads the OWL file from purl.obolibrary.org)
bfo.fetch_labels()
print(f"{len(bfo.label_list)} BFO classes loaded")
```

```python
# Build hierarchy DataFrame and save to data/reference/bfo_hierarchy.csv
df = bfo.get_hierarchy_df()
df
```

## Inspect the hierarchy

```python
# Top-level classes (classes that appear as Parent but never as Class)
all_classes  = set(df["Class"])
all_parents  = set(df["Parent class"])
roots = sorted(all_parents - all_classes)
print("BFO roots:", roots)
```

```python
# Verify the saved file is readable
loaded = dm.load_reference_csv("bfo_hierarchy")
print(f"Loaded {len(loaded)} rows from data/reference/bfo_hierarchy.csv")
loaded.head(10)
```
